```python
import matplotlib.pyplot as plt
import pandas as pd

df=pd.read_csv("C:\\Users\\Ashok S\\Downloads\\Car_Sales_CSV\\heart-disease.csv")
display(df)
```

```python
df.plot(kind="scatter",x="age",y="chol",c=df["target"],cmap="winter")
plt.show();
```

```python
fig,ax = plt.subplots(figsize=(15,6))
ax.bar(df["age"],df["trestbps"])
ax.set(title="Comparison of Age and Trestbps",xlabel="Age",ylabel="Trestbps")
mean_=(df["trestbps"].mean())
ax.axhline(mean_,c="black",linestyle="dotted")

plt.savefig("Comparing_Age_and_Trestbps")
plt.show();
```
